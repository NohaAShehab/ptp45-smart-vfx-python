```python
num1 = input("please num1 : ")
num2 = input("please enter num2: ")
num1 = int(num1)
num2 = int(num2)
operation = input("please enter operation: ")  # string
if operation == "+":
    res = num1 + num2
    print(res)
elif operation == '-':
    res = num1 - num2
    print(res)
elif operation == '*':
    res = num1 * num2
    print(res)
elif operation == '/':
    res = num1 / num2
    print(res)
else:
    print("----- not valid operation ")

```

<h5 style='color:yellow'> if you want to stop the loop ... break the loop # break </h>

```python
while True:
    num = input("please enter number ")
    print(num)
    if num=="done":
        break  # stop the loop
```

```python
# write a program to ask the user to enter 5 numbers and print their sum using while True
counter = 1
summ = 0
while True:
    num = input("please numbers")
    num = int(num)
    summ += num   # summ = summ + num
    counter +=1
    if counter == 6:
        break

print(summ)
```

```python
# write a program to ask the user to enter 7 numbers and print their multiplication using while True


result = 1
counter = 1

while True:
    usernum = input("please enter number: ")
    usernum = int(usernum)
    result *= usernum
    counter +=1
    if counter == 8:
        break

print(result)
```
