```python

# ask the user to enter anystring then covert it like below noha ---> NoHa

anystring = input("Please enter string: ")
index = 0
newstring = ''

while index < len(anystring):
    char = anystring[index]
    if index%2==0:
        newstring += char.upper()
    else:
        newstring += char.lower()
    index +=1

print("new string = "+ newstring)
```

```python

# replace a letter in a string  ==> e ===> a

message = input('Please enter message: ')
index = 0
newmessage = ""
while index < len(message):
    char = message[index]
    if char =="e":
        newmessage +='a'
    else:
        newmessage +=char

    index +=1

print(newmessage)
```

```python
### check if char exists in a string

# count vowels in anystring

ss = "hello from iti, ramdan kareem"
index= 0
no_of_vowels = 0
while index < len(ss):
    char = ss[index]

    if char=="a" or char=='e' or char=="i" or char=='o' or char=='u':
        no_of_vowels +=1

    index +=1

print(no_of_vowels)
```
