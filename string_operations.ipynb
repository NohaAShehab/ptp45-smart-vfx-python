# count vowels in any given word ?


```python
sentence = input("please enter any sentence")
index = 0
no_of_vowels = 0
while index < len(sentence):
    char = sentence[index]
    if char =='a'or char=='e' or char=='i' or char=='o' or char=='u':
        no_of_vowels +=1
    index +=1

print("total number of vowels in word = " + sentence +" "+ str(no_of_vowels))
```
